```python
reviews_vector_db = Chroma.from_documents(
    documents=documents,
    embedding=HuggingFaceEmbeddings(model_name="sentence-transformers/all-MiniLM-L6-v2"),
    persist_directory=REVIEWS_CHROMA_PATH,
)
```

```python
loader = CSVLoader(file_path=REVIEW_CSV, source_column="review")
documents = loader.load()

texts = [doc.page_content.replace("review:", "").strip() for doc in documents]

text_splitter = RecursiveCharacterTextSplitter(
    chunk_size=300,
    chunk_overlap=50,
    separators=["(?<=[.!?]) +"],
    is_separator_regex=True
)

split_docs = text_splitter.create_documents(texts)

# Embed and index this
reviews_vector_db.add_documents(split_docs)
```

```python

question = """Has anyone complained about communication with the hospital staff?"""
relevant_docs = reviews_vector_db.similarity_search(question, k=5)
```

```python
import random
import time
```

```python
def get_current_wait_time(hospital: str) -> int | str:
    """Dummy function to generate fake wait times"""

    if hospital not in ["A", "B", "C", "D"]:
        return f"Hospital {hospital} does not exist."
    time.sleep(1)  # Simulate a delay in getting the wait time
    return random.randint(1, 10000)  # Random wait time between 1 and 120 minutes

```
